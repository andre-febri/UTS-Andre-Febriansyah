```python
def tampilkan_nilai(data):
    print("Daftar Nilai:", data)

def tambah_nilai(data):
    nilai = int(input("Masukkan nilai baru: "))
    data.append(nilai)
    print("Data setelah ditambah:", data)

def hapus_nilai(data):
    nilai = int(input("Masukkan nilai yang ingin dihapus: "))
    if nilai in data:
        data.remove(nilai)
        print("Data setelah dihapus:", data)
    else:
        print("Nilai tidak ditemukan!")

def hitung_rata(data):
    rata = sum(data) / len(data)
    print("Rata-rata:", rata)

def nilai_max_min(data):
    print("Nilai tertinggi:", max(data))
    print("Nilai terendah:", min(data))

nilai_list = [70, 80, 90, 75, 85]

while True:
    print("\n=== MENU ===")
    print("1. Tampilkan nilai")
    print("2. Tambah nilai")
    print("3. Hapus nilai")
    print("4. Hitung rata-rata")
    print("5. Nilai tertinggi & terendah")
    print("6. Keluar")

    pilihan = input("Pilih menu: ")

    if pilihan == "1":
        tampilkan_nilai(nilai_list)
    elif pilihan == "2":
        tambah_nilai(nilai_list)
    elif pilihan == "3":
        hapus_nilai(nilai_list)
    elif pilihan == "4":
        hitung_rata(nilai_list)
    elif pilihan == "5":
        nilai_max_min(nilai_list)
    elif pilihan == "6":
        break
    else:
        print("Pilihan tidak valid!")
```
