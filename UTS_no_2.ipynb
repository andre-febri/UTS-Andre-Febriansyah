```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def insert_awal(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def insert_akhir(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return

        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = new_node

    def tampilkan(self):
        temp = self.head
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("None")

    def cari(self, key):
        temp = self.head
        while temp:
            if temp.data == key:
                print("Data ditemukan!")
                return
            temp = temp.next
        print("Data tidak ditemukan!")

    def hapus(self, key):
        temp = self.head

        if temp and temp.data == key:
            self.head = temp.next
            return

        prev = None
        while temp and temp.data != key:
            prev = temp
            temp = temp.next

        if temp is None:
            print("Data tidak ditemukan")
            return

        prev.next = temp.next


ll = LinkedList()

ll.insert_akhir(10)
ll.insert_akhir(20)
ll.insert_akhir(30)
ll.insert_akhir(40)

print("Linked List:")
ll.tampilkan()

print("Cari 30:")
ll.cari(30)

print("Hapus 20:")
ll.hapus(20)
ll.tampilkan()
```
