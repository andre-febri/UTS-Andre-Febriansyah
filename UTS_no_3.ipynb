```python
class Node:
    def __init__(self, judul, penyanyi):
        self.judul = judul
        self.penyanyi = penyanyi
        self.next = None
        self.prev = None

class CircularDoublyLinkedList:
    def __init__(self):
        self.head = None
        self.current = None

    def tambah_lagu(self, judul, penyanyi):
        new_node = Node(judul, penyanyi)


        if not self.head:
            self.head = new_node
            new_node.next = self.head
            new_node.prev = self.head
            self.current = new_node
        else:
            tail = self.head.prev
            tail.next = new_node
            new_node.prev = tail
            new_node.next = self.head
            self.head.prev = new_node

        print(f"Sukses Menambahkan: '{judul} - {penyanyi}' ke dalam playlist.")


    def hapus_lagu(self, judul):
        if not self.head:
            print("Gagal: Playlist kosong, tidak ada lagu yang bisa dihapus.")
            return

        curr = self.head
        found = False


        while True:
            if curr.judul.lower() == judul.lower():
                found = True
                break
            curr = curr.next
            if curr == self.head:
                break

        if not found:
            print(f"Gagal: Lagu dengan judul '{judul}' tidak ditemukan.")
            return


        if curr.next == curr and curr.prev == curr:
            self.head = None
            self.current = None
        else:

            curr.prev.next = curr.next
            curr.next.prev = curr.prev


            if curr == self.head:
                self.head = curr.next


            if curr == self.current:
                self.current = curr.next

        print(f"Sukses Menghapus: '{curr.judul} - {curr.penyanyi}' dari playlist.")
        del curr


    def tampilkan_playlist(self):
        if not self.head:
            print("\nPlaylist:")
            print("(Playlist Kosong)")
            return

        print("\nPlaylist:")
        curr = self.head
        idx = 1
        while True:
            print(f"{idx}. {curr.judul} - {curr.penyanyi}")
            curr = curr.next
            if curr == self.head:
                break


        if self.current:
            print("\nNow Playing:")
            print(f"{self.current.judul} - {self.current.penyanyi}")


            print("\nNext Lagu:")
            print(f"{self.current.next.judul} - {self.current.next.penyanyi}")


    def next_lagu(self):
        if not self.current:
            print("Gagal: Playlist kosong.")
            return
        self.current = self.current.next
        print(f"\n[Maju] Memutar Lagu Berikutnya:")
        print(f"Now Playing: {self.current.judul} - {self.current.penyanyi}")


    def prev_lagu(self):
        if not self.current:
            print("Gagal: Playlist kosong.")
            return
        self.current = self.current.prev
        print(f"\n[Mundur] Memutar Lagu Sebelumnya:")
        print(f"Now Playing: {self.current.judul} - {self.current.penyanyi}")


    def cari_lagu(self, judul):
        if not self.head:
            print("Gagal: Playlist kosong.")
            return

        curr = self.head
        idx = 1
        while True:
            if curr.judul.lower() == judul.lower():
                print(f"\nHasil Cari Lagu '{judul}':")
                print(f"-> Lagu Ditemukan di urutan ke-{idx}: {curr.judul} - {curr.penyanyi}")
                return
            curr = curr.next
            idx += 1
            if curr == self.head:
                break

        print(f"\nHasil Cari Lagu '{judul}':")
        print("-> Lagu Tidak Ditemukan.")



def main():
    playlist_cdll = CircularDoublyLinkedList()


    playlist_cdll.tambah_lagu("Hati-Hati di Jalan", "Tulus")
    playlist_cdll.tambah_lagu("Monokrom", "Tulus")
    playlist_cdll.tambah_lagu("Sial", "Mahalini")

    while True:
        print("\n===== MENU PLAYLIST =====")
        print("1. Tambah Lagu")
        print("2. Hapus Lagu")
        print("3. Tampilkan Playlist")
        print("4. Next Lagu")
        print("5. Prev Lagu")
        print("6. Cari Lagu")
        print("7. Keluar")

        pilihan = input("Pilih menu (1-7): ").strip()

        if pilihan == '1':
            print("\n--- Tambah Lagu Baru ---")
            judul = input("Masukkan Judul Lagu: ").strip()
            penyanyi = input("Masukkan Nama Penyanyi: ").strip()
            if judul and penyanyi:
                playlist_cdll.tambah_lagu(judul, penyanyi)
            else:
                print("Gagal: Judul dan Penyanyi tidak boleh kosong!")

        elif pilihan == '2':
            print("\n--- Hapus Lagu ---")
            judul = input("Masukkan Judul Lagu yang akan dihapus: ").strip()
            if judul:
                playlist_cdll.hapus_lagu(judul)
            else:
                print("Gagal: Judul tidak boleh kosong!")

        elif pilihan == '3':

            playlist_cdll.tampilkan_playlist()

        elif pilihan == '4':
            playlist_cdll.next_lagu()

        elif pilihan == '5':
            playlist_cdll.prev_lagu()

        elif pilihan == '6':
            print("\n--- Cari Lagu ---")
            judul = input("Masukkan Judul Lagu yang dicari: ").strip()
            if judul:
                playlist_cdll.cari_lagu(judul)
            else:
                print("Gagal: Input judul tidak valid!")

        elif pilihan == '7':
            print("\nKeluar dari program. Terima kasih!")
            break
        else:
            print("Pilihan tidak valid! Silakan masukkan angka 1 hingga 7.")

if __name__ == "__main__":
    main()
```
